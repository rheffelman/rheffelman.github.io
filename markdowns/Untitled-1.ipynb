```python
import itertools
letters = "abc"
max_len = 7

def strings_upto(letters, max_len, include_empty=False):
    start = 0 if include_empty else 1
    for r in range(start, max_len + 1):
        for tup in itertools.product(letters, repeat=r):
            yield ''.join(tup)

# Example:
gen = strings_upto(letters, max_len)
# iterate without storing everything
for s in gen:
    pass  # do work with s

# If you really need them all in memory (careful!):
all_strings = list(strings_upto(letters, max_len))
new_x = all_strings
affirmative = []
negative = []

for item in new_x:
    count = 0
    for i in range(len(item)):
        if i == 0 and len(item) != 1:
            if item[i] != item[i + 1]:
                count += 1
        elif i == len(item) - 1:
            if item[i] != item[i - 1]:
                count += 1
        elif i > 0 and i < len(item) - 1:
            if item[i] != item[i + 1] and item[i] != item[i - 1]:
                count += 1

    if count != 1:
        negative.append(item)
    if count == 1:
        affirmative.append(item)

# print(len(x))
# print(len(new_x))
# print(new_x)
# print(len(negative))
# print(len(affirmative))
# print(new_x)
negative.sort(key=len)
negative = negative[::-1]
for i in affirmative:
    print(f"{i} = (return)true")
```
